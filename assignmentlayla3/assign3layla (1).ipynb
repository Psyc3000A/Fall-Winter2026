```python
print([col for col in children.columns if "gender" in col.lower() or "sex" in col.lower()])
```

```python
children["Gender_code"] = (children["Gender"].astype(str).str.lower().isin(["male", "1"])).astype(int)
```

```python
print(children["Gender_code"].unique())
```

```python
children = children[["StandingHeightCm", "AgeInYearsAtScreening", "Gender_code"]].dropna()
```

```python
print(children.columns.tolist())
print(children.shape)
```

```python
from sklearn.linear_model import LinearRegression

m_age = LinearRegression(fit_intercept=False).fit(
    children[["AgeInYearsAtScreening"]],
    children["StandingHeightCm"]
)

m_age_const = LinearRegression().fit(
    children[["AgeInYearsAtScreening"]],
    children["StandingHeightCm"]
)

m_full = LinearRegression().fit(
    children[["AgeInYearsAtScreening", "Gender_code"]],
    children["StandingHeightCm"]
)
```

```python
print(m_age.coef_)
print(m_age_const.coef_)
print(m_full.coef_)
```

```python
children = children.sort_values("AgeInYearsAtScreening")

children["pred_age"] = m_age.predict(children[["AgeInYearsAtScreening"]])
children["pred_age_const"] = m_age_const.predict(children[["AgeInYearsAtScreening"]])
children["pred_full"] = m_full.predict(children[["AgeInYearsAtScreening", "Gender_code"]])
```

```python
children["Age_Gender"] = (
    children["AgeInYearsAtScreening"] * children["Gender_code"]
)

m_interaction = LinearRegression().fit(
    children[["AgeInYearsAtScreening", "Gender_code", "Age_Gender"]],
    children["StandingHeightCm"]
)

children["pred_interaction"] = m_interaction.predict(
    children[["AgeInYearsAtScreening", "Gender_code", "Age_Gender"]]
)
```

```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Model 1: Age without intercept
m_age = LinearRegression(fit_intercept=False).fit(
    children[["AgeInYearsAtScreening"]],
    children["StandingHeightCm"]
)

# Model 2: Age with intercept
m_age_const = LinearRegression().fit(
    children[["AgeInYearsAtScreening"]],
    children["StandingHeightCm"]
)

# Model 3: Age and gender
m_full = LinearRegression().fit(
    children[["AgeInYearsAtScreening", "Gender_code"]],
    children["StandingHeightCm"]
)

children = children.sort_values("AgeInYearsAtScreening")

children["pred_age"] = m_age.predict(children[["AgeInYearsAtScreening"]])
children["pred_age_const"] = m_age_const.predict(children[["AgeInYearsAtScreening"]])
children["pred_full"] = m_full.predict(children[["AgeInYearsAtScreening", "Gender_code"]])

# Interaction model
children["Age_Gender"] = children["AgeInYearsAtScreening"] * children["Gender_code"]

m_interaction = LinearRegression().fit(
    children[["AgeInYearsAtScreening", "Gender_code", "Age_Gender"]],
    children["StandingHeightCm"]
)

children["pred_interaction"] = m_interaction.predict(
    children[["AgeInYearsAtScreening", "Gender_code", "Age_Gender"]]
)
```
