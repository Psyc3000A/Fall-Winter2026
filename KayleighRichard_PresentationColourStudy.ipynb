```python
import pandas as pd

data = pd.read_csv("https://raw.githubusercontent.com/SchlossVRL/coloremoblues/master/T_BCP.csv")
display(data.head())
```

```python
import matplotlib.pyplot as plt

data.plot.scatter(x="L", y="Happy")
plt.xlabel("Colour lightness")
plt.ylabel("Average happiness rating")
plt.title("Colour Lightness and Happiness")
plt.show()
```
