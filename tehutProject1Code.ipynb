```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load your CSV file
df = pd.read_csv("your_file.csv")
```

```python
# Label the four experimental conditions
df["Condition"] = df.apply(
    lambda r: (
        "Baseline\n(Source 0, Emotion 0)" if r["SOURCE_IDENTITY"] == 0 and r["EMOTIONAL_INTENSITY"] == 0
        else "Emotion only\n(Source 0, Emotion 1)" if r["SOURCE_IDENTITY"] == 0
        else "Source only\n(Source 1, Emotion 0)" if r["EMOTIONAL_INTENSITY"] == 0
        else "Source + Emotion\n(Source 1, Emotion 1)"
    ),
    axis=1
)
```

```python
# Calculate sample-size-weighted mean IPC for each condition
summary = (
    df.groupby("Condition", sort=False)
      .apply(lambda x: (x["MEAN_IPC"] * x["SAMPLESIZE"]).sum() / x["SAMPLESIZE"].sum())
      .reset_index(name="Weighted Mean IPC")
)
```

```python
# Keep baseline first
order = [
    "Baseline\n(Source 0, Emotion 0)",
    "Emotion only\n(Source 0, Emotion 1)",
    "Source only\n(Source 1, Emotion 0)",
    "Source + Emotion\n(Source 1, Emotion 1)"
]
summary["Condition"] = pd.Categorical(summary["Condition"], categories=order, ordered=True)
summary = summary.sort_values("Condition")
```

```python
# Baseline mean for reference
baseline_mean = summary.loc[
    summary["Condition"] == "Baseline\n(Source 0, Emotion 0)",
    "Weighted Mean IPC"
].iloc[0]
```

```python
# Create bar graph
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=summary,
    x="Condition",
    y="Weighted Mean IPC",
    palette=["#7f8c8d", "#5dade2", "#af7ac5", "#ec7063"]
)
```

```python
# Add baseline reference line
plt.axhline(
    baseline_mean,
    color="black",
    linestyle="--",
    linewidth=1.2,
    label=f"Baseline mean = {baseline_mean:.2f}"
)
```

```python
# Add value labels
for bar in ax.patches:
    height = bar.get_height()
    ax.annotate(
        f"{height:.2f}",
        (bar.get_x() + bar.get_width() / 2, height),
        ha="center",
        va="bottom",
        xytext=(0, 4),
        textcoords="offset points"
    )
```

```python
plt.title("Mean IPC by Experimental Condition Compared With Baseline")
plt.xlabel("")
plt.ylabel("Sample-size-weighted Mean IPC")
plt.legend()
plt.tight_layout()
plt.show()
```
