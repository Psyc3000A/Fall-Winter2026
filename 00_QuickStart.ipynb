```python
import pandas as pd
import pingouin as pg
from io import StringIO
import matplotlib.pyplot as plt

# Create dataframe from inline CSV data
df = pd.read_csv(StringIO("""
Subject,Group,Time,Scores
1,Meditation,January,70
1,Meditation,June,80
2,Meditation,January,65
2,Meditation,June,75
3,Meditation,January,72
3,Meditation,June,78
4,Control,January,68
4,Control,June,69
"""))

# Keep Meditation group and exclude January
df = df.query("Group == 'Meditation' and Time != 'January'")

ax = pg.plot_paired(
    data=df,
    dv="Scores",
    within="Time",
    subject="Subject"
)

ax.set_title("Effect of meditation on school performance")
plt.show()
```
