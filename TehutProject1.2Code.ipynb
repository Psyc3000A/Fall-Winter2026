```python
import pandas as pd
datafile=pd.read_csv("https://raw.githubusercontent.com/Psyc3000A/Fall-Winter2026/refs/heads/main/tehutproject1-2.csv")
datafile.describe()
```

```python

import matplotlib.pyplot as plt

# Create a box plot of baseline IPC and post-interaction IPC
datafile.boxplot(column=["IPC_baseline", "IPC"])

plt.title("IPC Before and After Interaction")
plt.ylabel("Interpersonal Closeness (IPC)")
plt.xlabel("Measurement")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```
