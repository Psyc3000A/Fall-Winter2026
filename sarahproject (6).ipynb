```python
import pandas as pd

df = pd.read_csv('metaphor.csv')

df.head()
```

```python
summary = df[['anthroscore', 'competence', 'warmth']].describe().round(2)
summary
```

```python
import matplotlib.pyplot as plt

averages = df[['warmth', 'competence']].mean()

averages.plot(kind='bar')
plt.title('Average Warmth and Competence Scores')
plt.ylabel('Average Score')
plt.xlabel('Dimension')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
